### Tensors

A tensor is a data structure used to store numbers in PyTorch.

- **Scalar:** A single number, e.g. `torch.tensor(5)`
- **Vector:** A one-dimensional collection of numbers.
- **Matrix:** A two-dimensional collection of numbers.
- **Higher-dimensional tensor:** A collection with three or more dimensions, commonly used to represent batches of images.

The `shape` attribute tells us the dimensions of a tensor, while `dtype` tells us the data type of its values.

Tensors support mathematical operations and can run on CPUs or compatible GPUs.

Basic tensor operations

```python
import torch

# Create tensors
a = torch.tensor([[1., 2.], [3., 4.]])
b = torch.tensor([[5., 6.], [7., 8.]])

# Addition
print("Addition:\n", a + b)

# Element-wise multiplication
print("Element-wise multiplication:\n", a * b)

# Matrix multiplication
print("Matrix multiplication:\n", a @ b)

# Reshape a tensor
c = torch.arange(1, 7)
print("Original shape:", c.shape)
print("Reshaped tensor:\n", c.reshape(2, 3))

# Inspect properties
print("Shape:", a.shape)
print("Data type:", a.dtype)
```

### Autograd

Autograd is PyTorch's automatic differentiation system. It calculates gradients needed to train neural networks.

- `requires_grad=True`: tells PyTorch to track operations on a tensor for gradient calculation.
- `y.backward()`: calculates gradients using the chain rule.
- `x.grad`: stores the gradient of the output with respect to `x`.
- **Gradient:** tells us how a small change in a variable affects the output.

For example, if \(y=x^2+2x+1\), then \(\frac{dy}{dx}=2x+2\). At \(x=5\), the gradient is 12.

Autograd helps neural networks determine how to adjust their weights to reduce prediction errors.

Verify autograd

```python
import torch

x = torch.tensor(5.0, requires_grad=True)
y = x**2 + 2*x + 1

y.backward()

print("x:", x.item())
print("y:", y.item())
print("Gradient:", x.grad.item())

assert x.grad.item() == 12.0
print("Autograd test passed!")
```

### CPU and GPU in PyTorch

A CPU handles general-purpose computing, while a GPU can accelerate parallel mathematical operations used in deep learning.

PyTorch tensors can be moved to a compatible GPU using `.to(device)`. Google Colab provides access to a GPU when one is available and enabled.

The `torch.cuda.is_available()` function checks whether CUDA is available in the current runtime.

Test tensor execution on device

```python
import torch

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

x = torch.randn(1000, 1000, device=device)
y = x @ x

print("Device:", device)
print("Result shape:", y.shape)
print("Computation successful:", torch.isfinite(y).all().item())
```
