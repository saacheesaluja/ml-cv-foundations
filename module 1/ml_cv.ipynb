```python
import torch

x = torch.tensor([[1., 2.], [3., 4.]])
print(x.shape)
print(x.dtype)

y = x * 2
print(y)

z = torch.randn(3, 4)
print(z.shape)
```

```python
x = torch.tensor(5.0, requires_grad=True)
y = x**2 + 2*x + 1

y.backward()

print("Output:", y.item())
print("Gradient:", x.grad.item())
```
